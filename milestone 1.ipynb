```python
import pandas as pd

df = pd.read_csv("netflix_titles.csv")

# Remove all rows with missing values
df = df.dropna()

# Check missing values
print(df.isnull().sum())

# Save cleaned dataset
df.to_csv("netflix_cleaned.csv", index=False)

print("Dataset cleaned successfully!")
```

```python
df=df.drop_duplicates()
```

```python
print("rows and columns:")
print(df.shape)
```

```python
# Normalize text columns
df['country'] = df['country'].str.strip().str.title()
df['rating'] = df['rating'].str.strip().str.upper()
df['listed_in'] = df['listed_in'].str.strip().str.title()

# Convert date_added to datetime format
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Save normalized dataset
df.to_csv("netflix_normalized.csv", index=False)

print("Dataset normalized successfully!")
```

```python
# 1. Convert the comma-separated strings into actual Python lists
df['country'] = df['country'].str.split(', ')
df['listed_in'] = df['listed_in'].str.split(', ')

# 2. Explode 'country' so each country gets its own row
df = df.explode('country')

# 3. Explode 'listed_in' so each genre gets its own row
df = df.explode('listed_in')

# Let's see the truly normalized result
df[['country', 'rating', 'listed_in']].head()
```
