# Pandas

 Pandas is an open-source Python library used for data analysis and manipulation. It provides powerful tools for loading,cleaning,organizing,and processing structured data efficiently. It Organize data into tables called DataFrames and Save processed data back to files.

### In this project,Pandas helps to:

- Load the Netflix Dataset
- Check and handle missing values
- Remove duplicates records
- Normalize columns such as genre,rating,and country
- Save the cleaned dataset

```python
#import os module

import os

print(os.listdir("../dataset"))
```

```python
#Load Dataset

df = pd.read_csv("../dataset/netflix_titles.csv.csv")

print("Dataset Loaded Successfully")
df.head()
```

```python
#Dataset Shape

print("Rows and Columns:")
print(df.shape)
```

```python
#Missing Values

print("Missing Values:")
print(df.isnull().sum())
```

```python
#Handle Missing Values

df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna(df['rating'].mode()[0])
df['duration'] = df['duration'].fillna('Unknown')
df = df.dropna(subset=['date_added'])

print("Missing values handled")
```

```python
#Remove Duplicates

duplicates = df.duplicated().sum()
print("Duplicate rows before removal:", duplicates)

df = df.drop_duplicates()

print("Duplicate rows removed")
```

```python
#Normalize Country

df['country'] = df['country'].str.strip().str.title()

df['country'].head()
```

```python
#Normalize Rating

df['rating'] = df['rating'].str.strip().str.upper()

df['rating'].head()
```

```python
#Normalize Genre

df['listed_in'] = df['listed_in'].str.strip().str.title()

df['listed_in'].head()
```

```python
#Verify Dataset

print("Final Dataset Shape:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
df.to_csv("../dataset/netflix_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Conclusion

The Netflix dataset was successfully loaded and prepared for analysis. Missing values were handled, duplicate records were removed, and categorical features such as genre, rating, and country were normalized. The cleaned dataset is now consistent and ready for further analysis and visualization
